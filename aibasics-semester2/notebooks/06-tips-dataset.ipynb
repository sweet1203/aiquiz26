<div style="background:#f3f4f6;border-left:4px solid #9ca3af;border-radius:8px;padding:14px 18px;margin-top:16px;color:#4b5563;font-size:1.05em;line-height:1.65;">
<p style="margin:0 0 8px 0;font-weight:600;color:#374151;">💡 여기서 잠깐!</p>
<p style="margin:0 0 8px 0;color:#4b5563;"><strong style="color:#374151;">Q.</strong> 통계 정보를 확인할 때 <code style="background:#e5e7eb;padding:1px 6px;border-radius:4px;">sex</code>, <code style="background:#e5e7eb;padding:1px 6px;border-radius:4px;">smoker</code>, <code style="background:#e5e7eb;padding:1px 6px;border-radius:4px;">day</code>, <code style="background:#e5e7eb;padding:1px 6px;border-radius:4px;">time</code> 속성이 보이지 않는 이유는 무엇일까?</p><p style="margin:0;color:#4b5563;"><strong style="color:#374151;">A.</strong> 숫자형 데이터에 대한 통계를 계산하므로 범주형 데이터에 대한 통계는 기본적으로 제공하지 않는다. 범주형 데이터를 기계학습에 포함하기 위해서 <code style="background:#e5e7eb;padding:1px 6px;border-radius:4px;">get_dummies()</code> 함수를 사용해 0, 1로 부호화하여 숫자로 변경할 수 있다.</p>
</div>

<div style="background:#f0f9ff;border-left:3px solid #38bdf8;border-radius:6px;padding:14px 18px;margin-bottom:14px;">
<p style="margin:0 0 10px 0;font-size:1.3em;font-weight:700;color:#0369a1;">4. 범주형 속성을 숫자형으로 변경하기</p>

</div>

````python
import pandas as pd
dummyDf = pd.get_dummies(tips_df)  # 범주형을 숫자형으로 변환
dummyDf.head()
````

<div style="background:#f0f9ff;border-left:3px solid #38bdf8;border-radius:6px;padding:14px 18px;margin-bottom:14px;">
<p style="margin:0 0 10px 0;font-size:1.3em;font-weight:700;color:#0369a1;">5. 박스플롯으로 이상치 눈으로 확인하기</p>
</div>

````python
import seaborn as sns

# 박스플롯을 그리면, 상자 위아래로 벗어나 점으로 찍히는 값이 '이상치'입니다.
# 어려운 계산(IQR) 없이 그림으로 눈으로 확인해 봅니다.
sns.boxplot(data=tips_df[['total_bill', 'tip', 'size']])
````

<div style="background:#f3f4f6;border-left:4px solid #9ca3af;border-radius:8px;padding:14px 18px;margin-top:12px;color:#4b5563;line-height:1.65;">
💡 상자(박스)에서 <b>위·아래로 튀어나온 점</b>이 이상치입니다. total_bill 위쪽에 점이 몇 개 보이죠?
지금은 <b>눈으로 확인만</b> 하고 넘어갑니다. (실제로 제거할 때는 사분위수(IQR)로 경계를 계산하지만, 여기서는 생략합니다.)
</div>

<p style="margin:0 0 10px 0;font-size:1.3em;font-weight:700;color:#0369a1;">6. 히트맵으로 속성 간의 관계 파악하여 핵심 속성 찾기</p>
</div>

````python
# 숫자로 바꾼 데이터(dummyDf)의 상관관계를 히트맵으로 그려 핵심 속성을 찾습니다.
# 힌트: sns.heatmap( dummyDf.corr(numeric_only=True), annot=True )
````

<details><summary>💡 정답 보기</summary>

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))
sns.heatmap(dummyDf.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt='.2f')
plt.show()
```

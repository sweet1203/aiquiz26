# 1. 전통적 프로그래밍 vs 기계학습

| | 전통적 프로그래밍 | 기계학습 |
|---|---|---|
| 넣는 것 | 데이터 + **규칙** | 데이터 + **결괏값(정답)** |
| 나오는 것 | 결괏값 | **규칙**(모델) |
| 한 줄 | 사람이 규칙을 적어 줌 | 기계가 규칙(패턴)을 찾음 |


```python
# 서울 연평균 기온 예시 (수업용 작은 표)
기온 = pd.DataFrame({
    '연도': list(range(2015, 2025)),
    '평균기온': [13.4, 13.6, 13.1, 13.5, 13.8, 13.3, 14.0, 13.7, 14.2, 14.1]
})
display(기온)

# (가) 가장 더웠던 해는?  → 이미 있는 값을 찾으면 됨 (전통적)
가장더운 = 기온.loc[기온['평균기온'].idxmax()]
print('(가) 가장 더웠던 해:', int(가장더운['연도']), '년,', 가장더운['평균기온'], '도')
```

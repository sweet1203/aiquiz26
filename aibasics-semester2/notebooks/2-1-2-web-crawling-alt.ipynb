## 1. 인터넷 대신 파일을 읽기

원래는 `requests.get(주소)`가 HTML 글을 가져옵니다.  
막히면 **이미 받아 둔 HTML 파일**을 여는 것과 같습니다. 그다음 단계는 똑같습니다.


```python
with open('오늘_공지.html', encoding='utf-8') as 파일:
    html = 파일.read()

print('글자 수:', len(html))
print(html)
```

> **생각하기 1**
> 브라우저에서 이 파일을 열면 사람이 읽고, 여기서는 글자 덩어리가 나옵니다. 무엇이 같나요?

## 2. 필요한 칸만 꺼내기

`BeautifulSoup`은 HTML에서 원하는 태그를 찾습니다.  
인터넷에서 가져온 글이든, 파일에서 읽은 글이든 **쓰는 코드는 같습니다.**


```python
soup = BeautifulSoup(html, 'html.parser')

print('제목:', soup.find('h1').get_text())
print('날짜:', soup.find('p', class_='date').get_text())
print()
print('공지 목록')
for 칸 in soup.find_all('li', class_='item'):
    print('-', 칸.get_text())
```

## 3. (선택) 인터넷이 되면 원래 방식도 보기

집이나 핫스팟에서만 해보세요. 학교망에서 오류가 나도 **위 실습은 이미 끝난 것입니다.**


```python
import requests

주소 = 'https://example.com'
try:
    응답 = requests.get(주소, timeout=5)
    응답.raise_for_status()
    print('인터넷에서 가져왔습니다. 상태:', 응답.status_code)
    print(응답.text[:300])
except Exception as 오류:
    print('학교망이나 보안 프로그램이 막았습니다.')
    print('오류 이름:', type(오류).__name__)
    print('파일로 연습한 것과 같은 다음 단계(BeautifulSoup)는 그대로 쓰면 됩니다.')
```

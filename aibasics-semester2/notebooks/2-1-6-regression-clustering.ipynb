(다) 빠진 숫자를 구하는 문제와 (가) 무리 나누기 문제를 코드로 끝냅니다.  
**범주형 열을 숫자로 바꾸지 않으면 오류가 납니다.** 그 오류를 먼저 만나 보세요.

| 오늘 할 일 | 왜 필요한가? |
|---|---|
| 선형 회귀로 꽃받침 길이를 예측하고, k-평균으로 3무리를 만들기 | 분류 말고 예측·군집도 같은 표로 풀 수 있음을 확인합니다 |

**준비물:** `Iris1.csv` · scikit-learn  
**실행:** 위에서부터 ▶. 오류가 나도 당황하지 말고 다음 칸을 읽으세요.


```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

df = pd.read_csv('Iris1.csv', encoding='cp949')
df.head()
```

---

# 1. (다) 선형 회귀 — 꽃받침 길이 예측 (86~87쪽)

정답 y는 **숫자** `꽃받침 길이`입니다.


```python
X = df[['꽃받침 너비', '꽃잎 길이', '꽃잎 너비', '종류']]
y = df['꽃받침 길이']
X.head()
```

아래 칸은 **일부러 오류가 날 수 있습니다.** 메시지를 읽고 생각하기에 답을 적으세요.


```python
try:
    model = LinearRegression()
    model.fit(X, y)
    print('학습 성공')
except Exception as e:
    print(type(e).__name__)
    print(e)
```

> **생각하기 1**
> 오류 메시지에 `'버지니카'`가 보였다면, 컴퓨터가 무엇을 숫자로 바꾸지 못한 걸까요?
> →


```python
X숫자 = pd.get_dummies(X, columns=['종류'], dtype=int)
print('변환 후 열:', list(X숫자.columns))
display(X숫자.head())

X_train, X_test, y_train, y_test = train_test_split(X숫자, y, test_size=0.2, random_state=7)
reg = LinearRegression()
reg.fit(X_train, y_train)
pred = reg.predict(X_test)
print('결정계수 r2:', round(r2_score(y_test, pred), 3))
print('가중치:', [round(v, 3) for v in reg.coef_])
print('절편:', round(reg.intercept_, 3))
```

> **생각하기 2**
> `get_dummies`가 한 일을 한 줄로 적으세요. r2가 1에 가까우면 무엇을 뜻하나요?

# 2. (가) k-평균 군집 (88쪽)

종류 열을 **가리고** 숫자만으로 3무리를 만듭니다.


```python
군집용 = df[['꽃잎 길이', '꽃잎 너비']]
km = KMeans(n_clusters=3, random_state=7, n_init=10)
df['무리'] = km.fit_predict(군집용)

print('무리 번호 vs 실제 종류')
print(pd.crosstab(df['무리'], df['종류']))

plt.scatter(df['꽃잎 길이'], df['꽃잎 너비'], c=df['무리'], cmap='tab10', s=28)
cx, cy = km.cluster_centers_[:, 0], km.cluster_centers_[:, 1]
plt.scatter(cx, cy, marker='*', s=220, c='black', label='중심')
plt.xlabel('꽃잎 길이'); plt.ylabel('꽃잎 너비'); plt.legend()
plt.title('k-평균 (k=3)'); plt.show()
```

> **생각하기 3**
> 무리 번호(0,1,2)가 곧 '세토사'라는 이름은 아닙니다. 비지도학습이라서 생기는 차이는?

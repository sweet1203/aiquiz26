기계학습은 **사람이 특징을 골라** 주고, 딥러닝은 **스스로 특징을 찾습니다.**  
코랩의 Keras MNIST와 같은 6단계를, 교실에서 바로 도는 `digits`(8×8)로 체험합니다.

| 오늘 할 일 | 왜 필요한가? |
|---|---|
| 전처리 → 신경망(MLP) 학습 → 의사결정트리와 시험 점수 비교 | 훈련만 만점인 모델이 시험에서는 떨어질 수 있습니다 |

**준비물:** scikit-learn  
**실행:** 위에서부터 ▶.


```python
import pandas as pd
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import MinMaxScaler

digits = load_digits()
print('이미지 개수, 한 장 픽셀:', digits.data.shape)
print('예시 정답 10개:', digits.target[:10])
```

---

# 1. 전처리

교과서 MNIST: 28×28 → 784칸, 0~255를 255로 나눔, 레이블 원-핫.  
여기서는 8×8=64칸, 0~16을 0~1로 맞춥니다.


```python
X = digits.data
y = digits.target
X = MinMaxScaler().fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=7, stratify=y
)
print('훈련', X_train.shape, '시험', X_test.shape)
```

> 픽셀 값을 0~1로 나누는(정규화) 이유는 무엇일까요?
> →


---

# 2. 작은 신경망(MLP)과 의사결정트리


```python
mlp = MLPClassifier(hidden_layer_sizes=(100, 50), max_iter=400, random_state=7)
mlp.fit(X_train, y_train)
print('MLP   훈련', round(mlp.score(X_train, y_train), 3),
      '시험', round(mlp.score(X_test, y_test), 3))

tree = DecisionTreeClassifier(random_state=7)
tree.fit(X_train, y_train)
print('트리  훈련', round(tree.score(X_train, y_train), 3),
      '시험', round(tree.score(X_test, y_test), 3))
```

> **생각하기 2**
> 트리가 훈련에서 더 높아도 시험에서 더 낮다면, 교과서 101쪽의 어떤 말과 같나요? (과대적합)
> →


```python
# 시험 그림 하나 예측
import matplotlib.pyplot as plt
i = 0
plt.imshow(X_test[i].reshape(8, 8), cmap='gray_r')
plt.title(f'실제 {y_test[i]} / MLP 예측 {mlp.predict(X_test[i:i+1])[0]}')
plt.axis('off'); plt.show()
```

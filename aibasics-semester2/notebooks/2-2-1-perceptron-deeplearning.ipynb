# 1. 퍼셉트론 (91쪽)

출력 = 활성화함수( 입력1×가중치1 + 입력2×가중치2 + **편향** )

교과서 예: 가중치 1, 1 / 편향 −10 / 계단 함수(0 이상이면 1, 아니면 0)


```python
def 계단(x):
    return 1 if x >= 0 else 0

def 퍼셉트론(x1, x2, w1=1, w2=1, b=-10):
    s = x1 * w1 + x2 * w2 + b
    return s, 계단(s)

for x1, x2 in [(0, 0), (3, 3), (6, 6), (8, 3)]:
    s, y = 퍼셉트론(x1, x2)
    print(f'입력 ({x1}, {x2})  합={s:5}  출력={y}')
```

# 2. 단층 → 다층 → 딥러닝, 활성화 함수 (92쪽)

- 단층 퍼셉트론: 층이 하나  
- 다층: 은닉층이 있음  
- **딥러닝(DNN)**: 은닉층이 **여러 개**

| 위치 | 함수 |
|---|---|
| 은닉층 | 렐루 (ReLU, 음수는 0) |
| 이진 분류 출력 | 시그모이드 (0~1) |
| 다중 분류 출력 | 소프트맥스 (확률 합 1) |


```python
def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

x = np.array([-2.0, -0.5, 0.0, 2.0])
print('입력   ', x)
print('ReLU   ', relu(x))
print('sigmoid', np.round(sigmoid(x), 3))
```

> 은닉층에 시그모이드 대신 렐루를 많이 쓰는 이유로, 위 출력에서 보이는 차이를 한 줄로 적으세요.

# 3. 학습의 세 박자 (93~94쪽)

**순전파**(예측) → **오차 계산** → **역전파**(가중치 수정) 를 반복합니다.

손실의 예: 예측 문제의 MSE = 오차 제곱의 평균. 교과서 예시 값은 약 41.67입니다.


```python
실제 = np.array([90, 80, 70])
예측 = np.array([80, 70, 85])
오차 = 실제 - 예측
mse = np.mean(오차 ** 2)
print('오차:', 오차)
print('MSE :', round(float(mse), 2))
```

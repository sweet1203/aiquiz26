# 2. CNN = 합성곱층 + 풀링층 + 완전 연결층

1층 필터는 경계선, 2층은 눈·코, 3층은 얼굴처럼 **점점 큰 조각**을 봅니다.

체험: [CNN Explainer](https://poloclub.github.io/cnn-explainer/)


```python
# 교과서 105쪽 그림 Ⅱ-22: 4×4 입력에 3×3 필터 (합성곱)
이미지 = np.array([[1, 2, 3, 4],
                  [2, 1, 0, 1],
                  [3, 0, 1, 1],
                  [1, 4, 1, 1]])
필터 = np.array([[1, 0, 1],
                [1, 0, 1],
                [0, 1, 0]])

h, w = 이미지.shape
fh, fw = 필터.shape
결과 = np.zeros((h - fh + 1, w - fw + 1), dtype=int)
for i in range(결과.shape[0]):
    for j in range(결과.shape[1]):
        조각 = 이미지[i:i+fh, j:j+fw]
        결과[i, j] = int((조각 * 필터).sum())
print('합성곱 결과\n', 결과)

# 풀링(최댓값) — 교과서 105쪽 그림 Ⅱ-23
특징맵 = np.array([[5, 2, 3, 4],
                  [2, 1, 0, 1],
                  [3, 7, 9, 5],
                  [1, 4, 1, 8]])
풀링 = 특징맵.reshape(2, 2, 2, 2).max(axis=(1, 3))
print('풀링 결과\n', 풀링)
```

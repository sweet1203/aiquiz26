# 2. 잰말놀이 비교 (112쪽 흐름만)

원문과 '인식된 문장'이 얼마나 같은지 글자 단위로 봅니다.  
(실제 수업에서는 `speechrecognition` + 구글 인식, 언어 `ko-KR`.)


```python
원문 = '내가 그린 기린 그림은 긴 기린 그림이고 네가 그린 기린 그림은 짧은 기린 그림이다'
인식 = '내가 그린 기린 그림은 긴 기린 그림이고 내가 그린 기린 그림은 짧은 기린 그림이다'  # 일부러 한 글자 다름

print('길이 원문/인식:', len(원문), len(인식))
맞힌 = sum(a == b for a, b in zip(원문, 인식))
print('앞에서부터 같은 글자 수:', 맞힌, '/', min(len(원문), len(인식)))

표 = pd.DataFrame({'원문': list(원문), '인식': list(인식[:len(원문)])})
다른 = 표[표['원문'] != 표['인식']]
print('다른 위치')
display(다른.head(10))
```

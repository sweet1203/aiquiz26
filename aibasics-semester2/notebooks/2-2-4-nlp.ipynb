# 1. 글을 숫자로 (표 Ⅱ-11)

| 방법 | 예 | 특징 |
|---|---|---|
| 고유 번호 | 0, 1, 2… | 단순 인덱스 |
| 원-핫 | (1,0,0,0) | **희소**, 차원이 커짐 |
| 워드 임베딩 | (0.2, 0.4) | **밀집**, 비슷한 단어가 가깝게 |


```python
단어 = ['왕', '여왕', '사과', '바나나']
원핫 = pd.get_dummies(pd.Series(단어), dtype=int)
원핫.index = 단어
원핫
```

# 2. 단어 주머니 (빈도)

순서는 무시하고 **몇 번 나왔는지만** 셉니다.


```python
기사 = [
    '정부가 인공지능 교육을 확대한다',
    '인공지능 모델이 이미지를 분류한다',
]
vec = CountVectorizer()
X = vec.fit_transform(기사)
print(pd.DataFrame(X.toarray(), columns=vec.get_feature_names_out(), index=['기사1', '기사2']))
```

> **생각하기 2**
> '않는다 않는다'와 '않는다'처럼 순서가 뜻을 뒤집는 문장은 단어 주머니만으로 충분한가요?

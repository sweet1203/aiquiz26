## 0. 데이터 불러오기


```python
import pandas as pd

df = pd.read_csv('USA_Housing.csv')
print('행·열:', df.shape)
print('열 이름:', list(df.columns))
df.head()
```

> **힌트** 열 의미

# 1. 데이터 요약 정보 확인하기


```python
# 자료형 + non-null 개수
df.info()
```

```python
# 결측치 개수 (전부 0이면 '결측 없음')
print(df.isnull().sum())

# 수치형 / 문자열 개수 세기 힌트
num_cols = df.select_dtypes(include='number').columns
obj_cols = df.select_dtypes(exclude='number').columns
print('수치형 열 개수:', len(num_cols), list(num_cols))
print('문자열 열 개수:', len(obj_cols), list(obj_cols))
```

# 2. 독립/종속 변수 구분하기

- **종속 변수(y):** 맞히고 싶은 값 → `Price`
- **독립 변수(X):** 가격을 설명하는 나머지 숫자 열

> 만약 `Address` 열이 있다면 **문자열이라 회귀에 그대로 넣기 어렵습니다.** 빼고 시작하세요.


```python
# TODO: 아래 두 줄이 맞는지 확인하세요.
y = df['Price']

# Address가 있을 때만 제거 (없어도 에러 안 나게)
X = df.drop(columns=['Price'], errors='ignore')
if 'Address' in X.columns:
    X = X.drop(columns=['Address'])

print('독립 변수 열:', list(X.columns))
print('X:', X.shape, 'y:', y.shape)
```

---

# 3. 훈련/테스트 데이터 분할 (8:2)

`test_size=0.2` → 시험 20%, 훈련 약 80%


```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=7
)

# TODO: 네 개의 shape를 출력해 활동지 표에 옮기세요.
print('X_train', X_train.shape)
print('y_train', y_train.shape)
print('X_test ', X_test.shape)
print('y_test ', y_test.shape)
```

> **생각하기 2** 전체 행이 5000개라면, 대략 훈련 ____행 · 시험 ____행이어야 합니다. 출력과 맞나요?


---

# 4. 선형 회귀 모델 학습 및 예측


```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

# TODO: 훈련 데이터로 학습
model.fit(X_train, y_train)

# TODO: 시험 데이터로 예측
y_pred = model.predict(X_test)

# 앞에서 5개만 비교해 보기
비교 = pd.DataFrame({'실제': y_test.values[:5], '예측': y_pred[:5]})
비교
```

---

# 5. 모델 성능 평가 (MSE · R²)

| 지표 | 의미 | 좋은 쪽 |
|---|---|---|
| **MSE** (평균제곱오차) | 오차를 제곱해 평균 | **작을수록** 좋음 |
| **R²** (결정계수) | 설명을 얼마나 잘했는지 (최대 1에 가까움) | **1에 가까울수록** 좋음 |


```python
from sklearn.metrics import mean_squared_error, r2_score

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print('평균제곱오차(MSE):', round(mse, 2))
print('결정계수(R²):', round(r2, 4))
```

> **생각하기 3** R²가 0.9라면 "가격 변동의 약 90%를 모델이 설명한다"고 말할 수 있습니다.

```python
from pytrends.request import TrendReq
```

<div style="background:#fff7ed;border-left:3px solid #fb923c;border-radius:6px;padding:14px 18px;margin-bottom:14px;">
<p style="margin:0;color:#374151;line-height:1.6;font-size:1.05em;">지역을 한국(<code style="background:#fed7aa;padding:1px 6px;border-radius:4px;">hl='ko'</code>)으로, 시간대를 한국 시(<code style="background:#fed7aa;padding:1px 6px;border-radius:4px;">tz=540</code>)으로 설정한다.</p>
</div>

```python
trend = TrendReq(hl = 'ko', tz = 540)
keyword = ['독감']
```

```python
trend.build_payload(keyword, cat = 0, timeframe = '2023-01-01 2023-12-27', geo = 'KR', gprop = '')
```

```python
result = trend.interest_over_time()
result = result.reset_index()
result
```
